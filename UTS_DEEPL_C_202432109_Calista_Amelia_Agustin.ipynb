  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>

```python
#202432109_CalistaAmeliaAgustin

# Library untuk pengolahan data
import os
import re
import random
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Library PyTorch untuk pemodelan LSTM
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from torch.nn.utils.rnn import pack_padded_sequence

# Library untuk pembagian dan evaluasi data
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Menentukan seed agar hasil lebih konsisten
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

# Menggunakan CPU sesuai konfigurasi praktikum
device = torch.device("cpu")

print("Perangkat yang digunakan:", device)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432109_CalistaAmeliaAgustin

# 1. Memuat dataset
DATA_PATH = "ulasan_com.gojek.app.csv"

if not os.path.exists(DATA_PATH):
    from google.colab import files
    uploaded = files.upload()
    DATA_PATH = next(iter(uploaded))

df = pd.read_csv(DATA_PATH)

print("Ukuran dataset:", df.shape)
print("Nama kolom:", df.columns.tolist())

# 2. Memilih kolom ulasan dan rating
data = df[["Ulasan", "Rating"]].copy()

# Menghapus data yang kosong
data = data.dropna(subset=["Ulasan", "Rating"])

# Membersihkan format kolom
data["Ulasan"] = data["Ulasan"].astype(str).str.strip()
data["Rating"] = pd.to_numeric(data["Rating"], errors="coerce")
data = data.dropna(subset=["Rating"])

# Menghapus ulasan kosong
data = data[data["Ulasan"].str.len() > 0]

# 3. Mengubah rating menjadi kategori sentimen
# Rating 1-2 = Negatif, 3 = Netral, 4-5 = Positif
def rating_to_sentiment(rating):
    if rating <= 2:
        return 0
    elif rating == 3:
        return 1
    else:
        return 2

data["label"] = data["Rating"].apply(rating_to_sentiment)

class_names = ["Negatif", "Netral", "Positif"]

print("\nDistribusi sentimen:")
print(data["label"].value_counts().sort_index())

# 4. Melakukan case folding dan pembersihan teks
def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+|www\.\S+", " ", text)
    text = re.sub(r"[^a-z0-9\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

data["clean_text"] = data["Ulasan"].apply(clean_text)

# Menghapus teks yang kosong setelah dibersihkan
data = data[data["clean_text"].str.len() > 0].reset_index(drop=True)

# 5. Membagi data menjadi data training dan testing
X_train_text, X_test_text, y_train, y_test = train_test_split(
    data["clean_text"].tolist(),
    data["label"].to_numpy(),
    test_size=0.20,
    random_state=SEED,
    stratify=data["label"]
)

# 6. Membuat vocabulary dari data training
PAD_TOKEN = "<PAD>"
UNK_TOKEN = "<UNK>"

MAX_VOCAB = 15000
MAX_LEN = 60

token_counter = Counter()

for sentence in X_train_text:
    token_counter.update(sentence.split())

vocab = {
    PAD_TOKEN: 0,
    UNK_TOKEN: 1
}

for word, jumlah in token_counter.most_common(MAX_VOCAB - 2):
    vocab[word] = len(vocab)

# 7. Mengubah kata menjadi angka
def encode_text(text):
    tokens = text.split()

    token_ids = [
        vocab.get(token, vocab[UNK_TOKEN])
        for token in tokens
    ]

    token_ids = token_ids[:MAX_LEN]

    return token_ids

# 8. Melakukan padding agar panjang sequence sama
def pad_sequence(token_ids):
    return token_ids + [vocab[PAD_TOKEN]] * (MAX_LEN - len(token_ids))

X_train = torch.tensor(
    [pad_sequence(encode_text(text)) for text in X_train_text],
    dtype=torch.long
)

X_test = torch.tensor(
    [pad_sequence(encode_text(text)) for text in X_test_text],
    dtype=torch.long
)

y_train_tensor = torch.tensor(y_train, dtype=torch.long)
y_test_tensor = torch.tensor(y_test, dtype=torch.long)

# 9. Membuat TensorDataset dan DataLoader
train_dataset = TensorDataset(X_train, y_train_tensor)
test_dataset = TensorDataset(X_test, y_test_tensor)

train_loader = DataLoader(
    train_dataset,
    batch_size=256,
    shuffle=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=512,
    shuffle=False
)

print("\nJumlah data training:", len(train_dataset))
print("Jumlah data testing:", len(test_dataset))
print("Jumlah vocabulary:", len(vocab))
print("Panjang sequence:", MAX_LEN)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432109_CalistaAmeliaAgustin

# Mendefinisikan arsitektur jaringan LSTM
class SequenceModel(nn.Module):

    def __init__(self, vocab_size, embedding_dim=64, hidden_size=64):
        super(SequenceModel, self).__init__()

        # Mengubah indeks kata menjadi vektor numerik
        self.embedding = nn.Embedding(
            vocab_size,
            embedding_dim,
            padding_idx=0
        )

        # Mendefinisikan layer LSTM
        self.lstm = nn.LSTM(
            embedding_dim,
            hidden_size,
            batch_first=True
        )

        # Mengurangi risiko overfitting
        self.dropout = nn.Dropout(0.3)

        # Menghasilkan tiga kelas sentimen
        self.fc = nn.Linear(hidden_size, 3)

    def forward(self, x):

        # Menghitung panjang teks tanpa padding
        lengths = (x != 0).sum(dim=1).clamp(min=1)

        # Mengubah indeks kata menjadi embedding
        embedded = self.embedding(x)

        # Memproses sequence tanpa menghitung padding
        packed = pack_padded_sequence(
            embedded,
            lengths.cpu(),
            batch_first=True,
            enforce_sorted=False
        )

        # Memproses data menggunakan LSTM
        _, (hidden, cell) = self.lstm(packed)

        # Mengambil hidden state terakhir
        last_hidden = hidden[-1]

        # Menghasilkan skor prediksi sentimen
        output = self.fc(self.dropout(last_hidden))

        return output


# Membuat objek model
model = SequenceModel(
    vocab_size=len(vocab),
    embedding_dim=64,
    hidden_size=64
).to(device)

print(model)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432109_CalistaAmeliaAgustin

# 1. Inisiasi hyperparameter
epochs = 3
learning_rate = 0.001
hidden_size = 64

# 2. Menghitung bobot masing-masing kelas
class_counts = np.bincount(y_train, minlength=3)

class_weights = len(y_train) / (
    3.0 * np.maximum(class_counts, 1)
)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float32
).to(device)

# 3. Membuat model, fungsi loss, dan optimizer
model = SequenceModel(
    vocab_size=len(vocab),
    embedding_dim=64,
    hidden_size=hidden_size
).to(device)

criterion = nn.CrossEntropyLoss(weight=class_weights)

optimizer = optim.Adam(
    model.parameters(),
    lr=learning_rate
)

# 4. Menyimpan riwayat training loss dan akurasi
train_losses = []
test_accuracies = []

# 5. Menjalankan proses training
for epoch in range(epochs):

    model.train()
    total_loss = 0.0

    for batch_x, batch_y in train_loader:

        batch_x = batch_x.to(device)
        batch_y = batch_y.to(device)

        optimizer.zero_grad()

        outputs = model(batch_x)

        loss = criterion(outputs, batch_y)

        loss.backward()

        optimizer.step()

        total_loss += loss.item() * batch_x.size(0)

    # Menghitung rata-rata training loss
    epoch_loss = total_loss / len(train_loader.dataset)

    train_losses.append(epoch_loss)

    # Mengukur akurasi pada data testing
    model.eval()

    correct = 0
    total = 0

    with torch.no_grad():

        for batch_x, batch_y in test_loader:

            batch_x = batch_x.to(device)
            batch_y = batch_y.to(device)

            outputs = model(batch_x)

            predictions = torch.argmax(outputs, dim=1)

            correct += (predictions == batch_y).sum().item()
            total += batch_y.size(0)

    test_accuracy = correct / total

    test_accuracies.append(test_accuracy)

    print(
        f"Epoch {epoch + 1}/{epochs} "
        f"| Training Loss: {epoch_loss:.4f} "
        f"| Test Accuracy: {test_accuracy:.4f}"
    )
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432109_CalistaAmeliaAgustin

# 1. Menghasilkan prediksi model pada data testing
model.eval()

all_test_preds = []
all_test_labels = []

with torch.no_grad():

    for batch_x, batch_y in test_loader:

        batch_x = batch_x.to(device)

        outputs = model(batch_x)

        predictions = torch.argmax(outputs, dim=1)

        all_test_preds.extend(predictions.cpu().numpy())
        all_test_labels.extend(batch_y.numpy())


# 2. Plot Grafik Penurunan Loss Training terhadap Epoch
plt.figure(figsize=(10, 4))

plt.plot(
    range(1, len(train_losses) + 1),
    train_losses,
    color="#114F6A",
    linewidth=2,
    label="Training Loss"
)

plt.title("Grafik Penurunan Loss Selama Pelatihan")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.xticks(range(1, len(train_losses) + 1))
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()


# 3. Plot Perbandingan Data Aktual dan Hasil Prediksi
actual_counts = np.bincount(all_test_labels, minlength=3)
predicted_counts = np.bincount(all_test_preds, minlength=3)

x = np.arange(len(class_names))
width = 0.35

plt.figure(figsize=(12, 5))

plt.bar(
    x - width / 2,
    actual_counts,
    width,
    color="#114F6A",
    label="Data Aktual"
)

plt.bar(
    x + width / 2,
    predicted_counts,
    width,
    color="#1FA4B1",
    label="Hasil Prediksi"
)

plt.title("Perbandingan Data Aktual dan Hasil Prediksi Sentimen")
plt.xlabel("Kategori Sentimen")
plt.ylabel("Jumlah Data")
plt.xticks(x, class_names)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()


# 4. Menampilkan hasil evaluasi model
accuracy = accuracy_score(
    all_test_labels,
    all_test_preds
)

print("Akurasi Model:", f"{accuracy:.4f}")

print("\nClassification Report:")

print(
    classification_report(
        all_test_labels,
        all_test_preds,
        labels=[0, 1, 2],
        target_names=class_names,
        zero_division=0
    )
)


# 5. Menampilkan Confusion Matrix
cm = confusion_matrix(
    all_test_labels,
    all_test_preds,
    labels=[0, 1, 2]
)

plt.figure(figsize=(6, 5))

plt.imshow(cm)

plt.title("Confusion Matrix Model LSTM")
plt.xlabel("Hasil Prediksi")
plt.ylabel("Data Aktual")

plt.xticks(range(3), class_names)
plt.yticks(range(3), class_names)

for i in range(3):
    for j in range(3):
        plt.text(
            j,
            i,
            cm[i, j],
            ha="center",
            va="center"
        )

plt.colorbar(label="Jumlah Data")
plt.tight_layout()
plt.show()
```
